```python
import numpy as np
```

```python
sales = [0, 5, 155, 0, 518, 0, 1827, 616, 317, 325]
```

```python
sales_array = np.array(sales)
sales_array
```

```python
print(f'ndim: {sales_array.ndim})')
print(f'shape: {sales_array.shape})')
print(f'size: {sales_array.size})')
print(f'dtype: {sales_array.dtype})')
```

```python
#Array excercise
my_list = [x * 10 for x in range(1,11)]
my_list_array = np.array(my_list)
my_list_array
```

```python
print(f'ndim: {my_list_array.ndim})')
print(f'shape: {my_list_array.shape})')
print(f'size: {my_list_array.size})')
print(f'dtype: {my_list_array.dtype})')
```
