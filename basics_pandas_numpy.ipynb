```python
import numpy as np
prices  = np.array([5.99, 6.99, 22.99, 99.99, 4.99, 49.99])
```

```python
total = prices + 5
total
```

```python
import numpy as np


random_array = np.random.rand(12)

discount_pct = random_array[:6].reshape(6)   
pct_owed = 1 - discount_pct
final_owed = total + pct_owed
final_owed.round(2)
```
